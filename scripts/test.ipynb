```python
import os
import re

# Configuration
label_dir = r'C:\Users\ACER\nagarikta_project\dataset\labels\val'  # Folder containing your .txt annotation files
start_range = 211
end_range = 420
class_offset = 0

# Process files
updated_count = 0

# Get all files in the directory
for filename in os.listdir(label_dir):
    # Process only .txt files (YOLO/Object Detection labels)
    if filename.endswith(".txt"):
        # Find the number in the filename (e.g., 'image360' -> 360)
        match = re.search(r'\d+', filename)
        if match:
            file_num = int(match.group())
            
            # Use <= and >= to include both 360 and 630
            if start_range <= file_num <= end_range:
                file_path = os.path.join(label_dir, filename)
                
                new_lines = []
                with open(file_path, 'r') as f:
                    lines = f.readlines()
                    
                for line in lines:
                    parts = line.split()
                    if parts:
                        # parts[0] is the class ID (first integer on the line)
                        try:
                            old_class = int(parts[0])
                            new_class = old_class + class_offset
                            
                            # Replace the old class with the new class
                            parts[0] = str(new_class)
                            
                            # Reassemble the line with existing coordinates
                            new_lines.append(" ".join(parts) + "\n")
                        except ValueError:
                            # If for some reason the first part isn't a number, keep it as is
                            new_lines.append(line)
                
                # Overwrite the original file with the updated class IDs
                with open(file_path, 'w') as f:
                    f.writelines(new_lines)
                
                print(f"Shifted classes in: {filename}")
                updated_count += 1

print(f"\nSuccessfully updated {updated_count} files in the range [{start_range}-{end_range}].")
```
