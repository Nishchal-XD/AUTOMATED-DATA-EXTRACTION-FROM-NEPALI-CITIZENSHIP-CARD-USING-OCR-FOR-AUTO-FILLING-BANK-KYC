```python
# This code changes the name of all files in a specified directory

import os
import re

directory = r''
old_prefix = 'back'
new_prefix = 'image'
start_number = 360

# 1. Get all files starting with 'front'
files = [f for f in os.listdir(directory) if f.startswith(old_prefix)]

# 2. Sort files numerically (so front269 comes after front268, not front300)
def extract_number(f):
    match = re.search(r'\d+', f)
    return int(match.group()) if match else 0

files.sort(key=extract_number)

# 3. Rename files sequentially
for i, filename in enumerate(files):
    # Get the file extension (e.g., .jpg)
    extension = os.path.splitext(filename)[1]
    
    # Calculate new number
    current_new_num = start_number + i
    new_name = f"{new_prefix}{current_new_num}{extension}"
    
    # Full paths
    old_path = os.path.join(directory, filename)
    new_path = os.path.join(directory, new_name)
    
    # Rename
    os.rename(old_path, new_path)
    print(f"Renamed: {filename} -> {new_name}")

print(f"\nDone! Last file renamed was image{start_number + len(files) - 1}")
```

```python
# this code merges contents of one directory into another

import os
import shutil
from pathlib import Path

# Define your source and destination
source_dir = r'C:\Users\ACER\OneDrive\Documents\A_project'
dest_dir = r'C:\Users\ACER\nagarikta_project\dataset\labels\train'

# Ensure the destination directory exists
os.makedirs(dest_dir, exist_ok=True)

def merge_directories(src, dest):
    for item in os.listdir(src):
        s = os.path.join(src, item)
        d = os.path.join(dest, item)
        
        if os.path.isdir(s):
            # If it's a folder, recurse into it
            merge_directories(s, d)
        else:
            # If it's a file, copy it
            # Using shutil.copy2 preserves metadata (like timestamps)
            os.makedirs(os.path.dirname(d), exist_ok=True)
            shutil.copy2(s, d)

# Execute the merge
merge_directories(source_dir, dest_dir)
print(f"Contents of '{source_dir}' merged into '{dest_dir}' successfully.")
```

```python
# this code splits a dataset of images into training and testing sets

import os
import random
import shutil

# 1. Configuration
source_dir = r'C:\Users\ACER\nagarikta_project\dataset\images\train'  # Folder where image1 to image630 are located
train_dir = r'C:\Users\ACER\nagarikta_project\dataset\images\temp'
test_dir = r'C:\Users\ACER\nagarikta_project\dataset\images\val'
split_ratio = 0.8  # 80% for training

# Create destination folders
os.makedirs(train_dir, exist_ok=True)
os.makedirs(test_dir, exist_ok=True)

# 2. Identify all files
# This gets all files named image1...image630
all_files = [f for f in os.listdir(source_dir) if f.startswith('image')]

# 3. Shuffle the list randomly
random.seed(42) # Using a seed ensures you get the same 'random' split if you run it again
random.shuffle(all_files)

# 4. Calculate split point
split_point = int(len(all_files) * split_ratio)

train_files = all_files[:split_point]
test_files = all_files[split_point:]

# 5. Move (or Copy) the files
def move_files(file_list, destination):
    for filename in file_list:
        src = os.path.join(source_dir, filename)
        dst = os.path.join(destination, filename)
        shutil.copy2(src, dst) # Use copy2 to keep original files safe; use shutil.move to move them

move_files(train_files, train_dir)
move_files(test_files, test_dir)

# 6. Verification Report
print(f"Total images found: {len(all_files)}")
print(f"Images moved to Train: {len(train_files)} ({len(train_files)/len(all_files)*100:.1f}%)")
print(f"Images moved to Test: {len(test_files)} ({len(test_files)/len(all_files)*100:.1f}%)")
```

```python
# this code copies the entire content of one folder to another folder

import shutil
import os

def copy_folder_contents(source_dir, destination_dir):
    """
    Copies the entire content (files and subfolders) from source_dir to destination_dir.

    Args:
        source_dir (str): The path to the source directory.
        destination_dir (str): The path to the destination directory.
    """
    try:
        # Check if the destination directory already exists
        if os.path.exists(destination_dir):
            print(f"Destination directory '{destination_dir}' already exists. Deleting it first.")
            # Remove existing destination directory and its contents
            shutil.rmtree(destination_dir)

        # Copy the entire directory tree from source to destination
        shutil.copytree(source_dir, destination_dir)
        print(f"Successfully copied contents from '{source_dir}' to '{destination_dir}'.")

    except FileNotFoundError:
        print(f"Error: Source directory '{source_dir}' not found.")
    except Exception as e:
        print(f"An error occurred: {e}")

# --- Example Usage ---
# Define your source and destination directory paths
dir2 = r'C:\Users\ACER\nagarikta_project\dataset\labels\train'  # Source directory
dir1 = r'C:\Users\ACER\nagarikta_project\test'  # Destination directory

# Call the function to perform the copy
copy_folder_contents(dir1, dir2)
```
